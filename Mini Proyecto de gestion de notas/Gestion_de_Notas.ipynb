## SECCIÓN 1: Inicialización y Estructura de Datos

En esta sección inicializamos la lista principal `alumnos`, la cual almacenará diccionarios con el carnet, nombre y la pila de notas de cada estudiante registrado.

```python
# Inicialización de la base de datos en memoria
alumnos = []
```

## SECCIÓN 2: Registro de Estudiantes 
En esta sección definimos una función para registrar un nuevo estudiantes en la lista,validando previamente que el carnet no se encuentre duplicado.

```python
def registrar_estudiante():
    print("\n--- Registro de Estudiante ---")
    carnet = input("Ingrese el carnet: ").strip()

    # Validar que no exista el carnet
    for est in alumnos:
        if est["carnet"] == carnet:
            print("Ese carnet ya esta registrado.")
            return

    nombre = input("Ingrese el nombre completo: ").strip()
    alumnos.append({"carnet": carnet, "nombre": nombre, "pila_notas": []})
    print("Estudiante guardado correctamente.")
```

## SECCIÓN 3: Agregar notas a estudiante
En esta seccion definimos para asignar  calificaciones a  un estudiante registrado ;agregando la nueva nota a su pila de notas previa validando del rango de la nota y existencia del carnet.

```python

def agregar_nota():
    if not alumnos:
        print("\nNo hay estudiantes registrados todavia.")
        return

    print("\nEstudiantes registrados:")
    for est in alumnos:
        print("Carnet:", est["carnet"], "| Nombre:", est["nombre"])

    carnet = input("\nIngrese el carnet del alumno para agregar nota: ").strip()

    for est in alumnos:
        if est["carnet"] == carnet:
            try:
                nota = float(input("Ingrese la nota: "))
                if nota < 0 or nota > 100:
                    print("La nota debe estar entre 0 y 100.")
                    return
                # Insertamos la nota al final (comportamiento de Pila)
                est["pila_notas"].append(nota)
                print("Nota agregada con exito.")
            except ValueError:
                print("Error: Debe ingresar un numero valido.")
            return

    print("El carnet ingresado no existe en el sistema.")

```

## SECCIÓN 4: Eliminar la ultima nota.
Definimos la funcion para eliminar la calificacion mas reciente de un  estudiante.

```python
def eliminar_ultima_nota():
    if not alumnos:
        print("\nNo hay estudiantes registrados.")
        return

    carnet = input("\nIngrese el carnet del alumno: ").strip()

    for est in alumnos:
        if est["carnet"] == carnet:
            pila = est["pila_notas"]
            if not pila:
                print("El alumno no tiene notas para eliminar.")
                return

            # Sacamos la ultima nota ingresada (POP)
            nota_eliminada = pila.pop()
            print("Se elimino la ultima nota ingresada:", nota_eliminada)
            return

    print("El carnet ingresado no existe en el sistema.")
```

## SECCIÓN 5: Consultar historial de notas 
Esta seccion se encarga  de visualizar las notas de un estudiante especifico,mostrando su historial comenzandon por la ultima nota ingresada.

```python
def mirar_registro_de_nota():
    if not alumnos:
        print("\nNo hay estudiantes registrados.")
        return

    print("\nEstudiantes registrados:")
    print("-----------------------------------")
    for est in alumnos:
        print("Carnet:", est["carnet"], "| Nombre:", est["nombre"])
    print("-----------------------------------")

    carnet = input("Ingrese el carnet del alumno a consultar: ").strip()

    for est in alumnos:
        if est["carnet"] == carnet:
            pila = est["pila_notas"]
            if not pila:
                print("El alumno", est["nombre"], "no tiene notas guardadas.")
                return

            print("\nNotas de:", est["nombre"], "[", est["carnet"], "]")
            print("-----------------------------------")

            # Muestra primero la ultima nota registrada
            num_nota = len(pila)
            for nota in reversed(pila):
                if num_nota == len(pila):
                    print("Nota", num_nota, ":", nota, "(Ultima ingresada)")
                else:
                    print("Nota", num_nota, ":", nota)
                num_nota -= 1

            # Calculo de la suma y promedio
            suma_notas = 0
            for n in pila:
                suma_notas += n

            promedio = suma_notas / len(pila)

            print("-----------------------------------")
            print("Total de notas:", len(pila))
            print("Promedio actual:", round(promedio, 2))
            return

    print("El carnet ingresado no existe en el sistema")
```

## SECCIÓN 6: Menu principal e interacciones  con el usuario .
En esta seccion definimos la funcion del menu unteractivo ,la cual despliega las opciones diponibles en la consola y cordina el flujo del programa.

```python
def menu_principal():
    while True:
        print("\n========== SISTEMA DE NOTAS ==========")
        print("1. Registrar estudiante")
        print("2. Agregar nota a estudiante")
        print("3. Eliminar ultima nota (Deshacer)")
        print("4. Consultar notas y promedio")
        print("5. Salir")

        opcion = input("Seleccione una opcion: ").strip()

        if opcion == "1":
            registrar_estudiante()
        elif opcion == "2":
            agregar_nota()
        elif opcion == "3":
            eliminar_ultima_nota()
        elif opcion == "4":
            mirar_registro_de_nota()
        elif opcion == "5":
            print("Saliendo del programa...")
            break
        else:
            print("Opcion invalida, intente de nuevo.")


# Ejecucion del programa
menu_principal()
```

Conclusión:El programa implementa la gestión de calificaciones en Python combinando estructuras de datos fundamentales (listas y diccionarios) con el comportamiento de una Pila (LIFO) mediante las operaciones append() y pop(). Garantiza la integridad de los datos mediante validaciones de duplicados y rangos numéricos, ofreciendo un control interactivo completo desde la consola.
